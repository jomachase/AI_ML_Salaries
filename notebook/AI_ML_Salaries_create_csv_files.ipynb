# Load libraries

```python
#Libraries
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import statistics
import scipy.stats as st
import os
```

# Read configuration file

```python
import yaml
 
try:
    with open("../config.yaml", "r") as file:
        config = yaml.safe_load(file)
except:
    print("Yaml configuration file not found!")
```

# Read data

```python
df = pd.read_csv(config['output_data']['file'])
df = df.drop_duplicates()
df.head()
```

# Preparing file for the relational database

## Experience file

```python
import pandas as pd

# Create the experience DataFrame
experience_df = pd.DataFrame({
    "experience_id": [i for i in range(1, df['Experience_Level'].nunique() + 1)],
    "name": df['Experience_Level'].unique()
})

# Trim spaces from all string columns in experience_df
experience_df = experience_df.applymap(lambda x: x.strip() if isinstance(x, str) else x)

# Display the cleaned DataFrame
display(experience_df)

# Save the cleaned DataFrame to CSV
experience_df.to_csv(config['output_data']['file1'], index=False)
```

```python
experience_df = pd.DataFrame({"experience_id": [ i for i in range(1, df['Experience_Level'].nunique() + 1)], "name": df['Experience_Level'].unique()})
display(experience_df)
experience_df.to_csv(config['output_data']['file1'], index=False)
```

## Currencies file

```python
currency_df = pd.DataFrame({"currency_id": [ i for i in range(1, df['Currency'].nunique() + 1)], "name": df['Currency'].unique()})
display(currency_df)
currency_df.to_csv(config['output_data']['file2'], index=False)
```

## Company size file

```python
company_size_df = pd.DataFrame({"size_id": [ i for i in range(1, df['Company_Size'].nunique() + 1)], "name": df['Company_Size'].unique()})
display(company_size_df)
company_size_df.to_csv(config['output_data']['file3'], index=False)
```

## Employment type file 

```python
employment_type_df = pd.DataFrame({"employment_id": [ i for i in range(1, df['Employment_Type'].nunique() + 1)], "name": df['Employment_Type'].unique()})
display(employment_type_df)
employment_type_df .to_csv(config['output_data']['file4'], index=False)
```

## Remote File

```python
remote_df = pd.DataFrame({"remote_id": [ i for i in range(1, df['Remote_Amount'].nunique() + 1)], "percentage": df['Remote_Amount'].unique()})
display(remote_df)
remote_df.to_csv(config['output_data']['file5'], index=False)
```

## Positions

```python
positions_df = pd.DataFrame({"position_id": [ i for i in range(1, df['Position'].nunique() + 1)], "position": df['Position'].unique()})
display(positions_df)
positions_df.to_csv(config['output_data']['file6'], index=False)
```

## Countries

```python
countries_df = pd.DataFrame({"country_id": [ i for i in range(1, df['Country'].nunique() + 1)], "name": df['Country'].unique()})
display(countries_df)
countries_df.to_csv(config['output_data']['file7'], index=False)
```

## Job id

```python
job = df.copy()
job['job_id'] = [ i for i in range(1, job.shape[0]+1)]
job = job.iloc[:, [-1] + [i for i in range(job.shape[1]-1)]]
job.head()
```

# Combining tables

```python
job = job.merge(remote_df, how="left", left_on="Remote_Amount", right_on="percentage")
job = job.loc[:, [col for col in job.columns if col not in ["Remote_Amount", "percentage"]]] 
job.head()
```

```python
job = job.merge(company_size_df, how="left", left_on="Company_Size", right_on="name")
job = job.loc[:, [col for col in job.columns if col not in ["Company_Size", "name"]]]
job.head()
```

```python
job = job.merge(currency_df, how="left", left_on="Currency", right_on="name")
job = job.loc[:, [col for col in job.columns if col not in ["Currency", "name"]]]
job.head()
```

```python
job = job.merge(experience_df, how="left", left_on="Experience_Level", right_on="name")
job = job.loc[:, [col for col in job.columns if col not in ["Experience_Level", "name"]]]
job.head()
```

```python
job = job.merge(employment_type_df, how="left", left_on="Employment_Type", right_on="name")
job = job.loc[:, [col for col in job.columns if col not in ["Employment_Type", "name"]]]
job.head()
```

```python
job = job.merge(positions_df, how="left", left_on="Position", right_on="position")
job = job.loc[:, [col for col in job.columns if col not in ["Position", "position"]]]
job.head()
```
